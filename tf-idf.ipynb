```python
#data cleaning and preprocessing already once done in BagOfWords tutorial
import pandas as pd
msg = pd.read_csv("datasets/spam.csv", encoding='latin1')
msg.drop(columns=['Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4'], inplace=True)
msg.rename(columns = {'v1': 'label', 'v2': 'message'}, inplace=True)
print(msg)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import re
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
corpus = []
```

```python
print(msg['message'])
```

```python
for i in range(len(msg)):
    review = re.sub('[^a-zA-Z]', ' ', msg['message'][i])
    review = review.split()
    review = [ps.stem(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
```

```python
print(corpus)
```

```python
tfidf = TfidfVectorizer(input='content', lowercase=True, analyzer='word', stop_words='english', max_features=100, ngram_range = (1,2))
X = tfidf.fit_transform(corpus).toarray()
```

```python
print(tfidf.vocabulary_)
```

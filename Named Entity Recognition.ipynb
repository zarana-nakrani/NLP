## Named Entity Recognition
### what is NER and why we do it?
NER is basically describing role of each word in the given text. This helps in explaining and training the model with what each word means

```python
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('maxent_ne_chunker')
```

```python
nltk.download('words')
```

```python
paragraph = """
Arthur Samuel of IBM developed a computer program for playing checkers in the 1950s. 
Since the program had a very small amount of computer memory available, 
Samuel initiated what is called alpha-beta pruning. 
His design included a scoring function using the positions of the pieces on the board. 
The scoring function attempted to measure the chances of each side winning. 
The program chooses its next move using a minimax strategy, 
which eventually evolved into the minimax algorithm.
"""

"""
Person : Arthur Samuel
Organization: IBM
Date: 1950s
// other things that can be recognized can be Time, Money, Place, Percent and many more.
"""
```

```python
sentences = nltk.sent_tokenize(paragraph)
sentences
```

```python
for sentence in sentences:
    words = nltk.word_tokenize(sentence)
    words = [word for word in words if word not in set(stopwords.words("english"))]
    tag_elements = nltk.pos_tag(words)
    nltk.ne_chunk(tag_elements).draw()
```

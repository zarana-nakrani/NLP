### Vocabulary {unique words} for One-hot Encoding
find the unique words in the given data
So for each word in the sentence, we put 1 for the word we are representing and all other words will be zero

### Disadvantages of One-hot Encoding
1. Sparse Matrix (A Matrix that has so many 1's and 0's) leads to Overfitting
2. Out of Vocabulary - meaning if in the test dataset there are words that are not present in the training dataset than model will not be able to represent that words in the form vector
3. No sensitive meaning is captured
   - Eg : we have 3 words in vocablury --> {food, pizza, burger}
   - One hot encoding of each word will be like:
     - food: [1 0 0]
     - pizza: [0 1 0]
     - burger: [0 0 1]
    so when we represent this in a 2-d graph all the words will be at the same distance from the center and hence model will not be able to capture the relation between this words
